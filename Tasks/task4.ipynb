**Physical sources of QRNG quantum entropy:**

- Photonic QRNGs : based on photon properties like polarization, phase or arrival time
- Quantum Tunneling QRNGs: based on electron tunneling through a barrier which happens randomly
- vaccum fluctuation QRNGs: based on unpredictable variations in elctromagnetic energy
