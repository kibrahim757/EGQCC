# Research Component

QBER "Quantum Bit error rate" is a measure of the error rate in bit transmission in BB84. as the protocol works as follows 

1. Alice sends encoded qubits in two bases Z basis $|0\rangle$ , $|1\rangle$ or $|+\rangle$, $|-\rangle$ to Bob.
2. Bob measures these qubits at random chosen bases, if he chooses the same bases the result will be correct , if not it will be random.
3. Bob and Alice share the basis used without revealing the actual values, compare their results and use the shared keys.
4. If Eve is intercpeting the transmission, the error rate will be higher than expected.
5. for one way communications the theortical threshold is 12.5% , however a more accurate QBER is 11%  to accomodate for channel noise.

As for Quantum Temporal Authentication, it can encode information in a way that is linked to a point in time. So if an attacker trys to replay or forward the communication, it attempts to insert that information at a different position which will be detected as an anomaly.

Using QuNetSim to build a simulation of a 3-node network (Alice, Bob, and Eve).

## Alice sends to Bob

```python
from qunetsim.components import Host, Network
from qunetsim.objects import Message, Qubit, Logger
from qunetsim.backends import EQSNBackend
Logger.DISABLED = True

# create the EQSN backend object
backend = EQSNBackend()


def protocol_1(host, receiver):
    # Here we write the protocol code for a host.
    #for i in range(5):
    #    s = 'Hi {}.'.format(receiver)
    #    print("{} sends: {}".format(host.host_id, s))
    #    host.send_classical(receiver, s, await_ack=True)
    for i in range(5):
        q = Qubit(host)
        q.X()
        print("{} sends qubit in the |1> state".format(host.host_id))
        host.send_qubit(receiver, q, await_ack=True)


def protocol_2(host, sender):
    # Here we write the protocol code for another host.
    #for i in range(5):
    #    sender_message = host.get_classical(sender, wait=5, seq_num=i)
    #    print("{} Received classical: {}".format(host.host_id, sender_message.content))
    for i in range(5):
        q = host.get_qubit(sender, wait=10)
        m = q.measure()
        print("{} measured: {}".format(host.host_id, m))


def main():
    network = Network.get_instance()
    nodes = ['Alice', 'Bob', 'Eve']
    network.generate_topology(nodes, 'linear')
    network.start(nodes)

    host_alice = network.get_host('Alice')
    host_bob = network.get_host('Bob')
    host_eve = network.get_host('Eve')

    t1 = host_alice.run_protocol(protocol_1, (host_bob.host_id,))
    t2 = host_bob.run_protocol(protocol_2, (host_alice.host_id,), blocking=True)
    network.stop(True)


if __name__ == '__main__':
    main()
```
